# ✅ Q11. Write a function calculate_HRR(max_bpm, resting_bpm, intensity=0.7) that returns Heart Rate Reserve zone.


```python
def calculate_HRR(max_bpm, resting_bpm, intensity=0.7):   #Function to calculate HRR using Karvonen Formula
  hrr = max_bpm - resting_bpm
  target_hr = resting_bpm + hrr * intensity
  return round(target_hr,2)
tar_hr = calculate_HRR(180, 60)
print(tar_hr)
```

# ✅ Q17. Write a recursive function sum_n(n) to return sum of first n natural numbers. Call sum_n(100).


```python
def sum_n(n):   #Function to return sum of first n natural numbers
  if n == 0:
    return 0
  else:
    return n + sum_n(n-1) #Recursive call
print(sum_n(100)) #Calling the function and print what it returns
```

# ✅ Q18. Write a recursive function factorial(n) to return n!. Call factorial(6).


```python
def factorial(n):   #Function for calculating factorial
  if n == 0 :
    return  1
  return  n * factorial(n-1)  #Recursive call
print(factorial(6))   #Calling the function and print what it returns
```

# ✅ Q19. Write a function mean_std(array) that returns mean and standard deviation using NumPy. Test: import numpy as np
# data = np.array([10,20,30,40,50])
# print(mean_std(data))


```python
def mean_std(array): #Function to return mean and standard deviation
  mean = np.mean(array)
  std = np.std(array)
  return mean, std  #Returns values in numPy datatype
data = np.array([10,20,30,40,50])
print(mean_std(data))
mean_value = float(mean_std(data)[0])
std_value = float(mean_std(data)[1])
print("In normal format:",mean_value,",", std_value)
```
